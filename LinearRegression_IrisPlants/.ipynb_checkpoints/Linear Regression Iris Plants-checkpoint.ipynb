The famous Iris database, first used by Sir R.A. Fisher. The dataset is taken from Fisher’s paper. Note that it’s the same as in R, but not as in the UCI Machine Learning Repository, which has two wrong data points.

This is perhaps the best known database to be found in the pattern recognition literature. Fisher’s paper is a classic in the field and is referenced frequently to this day. (See Duda & Hart, for example.) The data set contains 3 classes of 50 instances each, where each class refers to a type of iris plant. One class is linearly separable from the other 2; the latter are NOT linearly separable from each other.

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_iris
import numpy as np
```

```python
# Iris dataset
iris = load_iris()

df = pd.DataFrame(iris.data, columns = iris.feature_names)
```

```python
df.head()
```

```python
# Set the x-values to the petal length:
X = df[["petal length (cm)"]]
# Y-values are petal width:
y = df[["petal width (cm)"]]
```

```python
plt.scatter(X, y, alpha=0.4)

# Plot line here:
line_fitter = LinearRegression()
line_fitter.fit(X, y)
width_predict = line_fitter.predict(X)
plt.plot(X, width_predict)

plt.title("Iris Plants Dataset")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()
plt.clf()
```

```python
# Predict new y values from new X values
plt.scatter(X, y, alpha=0.4)

# Plot line here:
line_fitter = LinearRegression()
line_fitter.fit(X, y)
width_predict = line_fitter.predict(X)
plt.plot(X, width_predict)

plt.title("Iris Plants Dataset")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

X_petals = np.array(range(7, 16))
X_petals = X_petals.reshape(-1,1)

predicted_width = line_fitter.predict(X_petals)
plt.plot(X_petals, predicted_width)

plt.show()
plt.clf()
```

```python
# Set the x-values to the sepal length:
S = df[["sepal length (cm)"]]
# Y-values are petal length:
l = df[["petal length (cm)"]]
```

```python
plt.scatter(S, l, alpha=0.4)

# Plot line here:
line_fitter = LinearRegression()
line_fitter.fit(S, l)
new_length_predict = line_fitter.predict(S)
plt.plot(S, new_length_predict)

plt.title("Iris Plants Dataset")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.show()
plt.clf()
```

```python
plt.scatter(S, l, alpha=0.4)

# Plot line here:
line_fitter = LinearRegression()
line_fitter.fit(S, l)
new_length_predict = line_fitter.predict(S)
plt.plot(S, new_length_predict)

plt.title("Iris Plants Dataset")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

X_longer = np.array(range(8, 11))
X_longer = X_longer.reshape(-1,1)

predicted_sepal = line_fitter.predict(X_longer)

plt.plot(X_longer, predicted_sepal)

plt.show()
plt.clf()
```
