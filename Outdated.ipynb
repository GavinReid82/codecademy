```python
months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]
```

```python
def main():
    date = input("Enter a date: ")
    date_fixed = date_split(date)
    print(date_fixed)
```

```python
def date_split(date):
    for i in date:
        if i == "/":
            m, d, y = map(int, date.split("/"))
            return f"{y}/{m:02}/{d:02}"
        elif i == " ":
            month, day, y = date.split(" ")
            d = int(day)
            if month in months:
                month_index = months.index(month) + 1
                return f"{y}/{month_index:02}/{d:02}"
```

```python
main()
```
